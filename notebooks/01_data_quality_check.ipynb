# 01 Data Quality Check

This notebook performs the initial data quality and structure checks for the Global Healthcare & Pharma Analytics project.

The goal of this notebook is to verify that all datasets are correctly loaded, inspect their structure, check missing values, validate data types, convert date columns, and identify duplicate records before starting exploratory analysis.

## Step 1: Import Required Libraries

In this step, we import the core Python libraries used for data analysis. Pandas is used for working with tabular data, while NumPy supports numerical operations. Display settings are adjusted to make the notebook outputs easier to inspect.

```python
import pandas as pd
import numpy as np
import os
```

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)
```

## Step 2: Check Available Project Files

Before loading the datasets, we check the files available in the current working directory. This confirms that all required CSV files are stored in the same folder as the notebook.

## Step 3: Load CSV Files

In this step, each CSV file is loaded into a separate Pandas DataFrame. These DataFrames will be used throughout the project for analysis.

```python
clinical_trials = pd.read_csv("clinical_trials.csv")
disease_burden = pd.read_csv("disease_burden.csv")
biotech_funding = pd.read_csv("biotech_funding.csv")
drug_approvals = pd.read_csv("drug_approvals.csv")
pharma_financials = pd.read_csv("pharma_companies_financials.csv")
```

## Step 9: Check Dataset Dimensions

This step checks the number of rows and columns in each dataset. Understanding dataset size helps validate whether the files were loaded as expected and gives an overview of the available data volume.

```python
print("clinical_trials:", clinical_trials.shape)
print("disease_burden:", disease_burden.shape)
print("biotech_funding:", biotech_funding.shape)
print("drug_approvals:", drug_approvals.shape)
print("pharma_financials:", pharma_financials.shape)
```

## Step 10: Inspect Column Names

This step lists the column names in each dataset. Reviewing column names is necessary before writing analysis code, joining datasets, or creating custom KPIs.

```python
print("clinical_trials columns:")
print(clinical_trials.columns.tolist())

print("\ndisease_burden columns:")
print(disease_burden.columns.tolist())

print("\nbiotech_funding columns:")
print(biotech_funding.columns.tolist())

print("\ndrug_approvals columns:")
print(drug_approvals.columns.tolist())

print("\npharma_financials columns:")
print(pharma_financials.columns.tolist())
```

## Step 11: Check Missing Values

This step checks whether any columns contain missing values. Missing values can affect calculations, aggregations, and visualizations, so they must be identified before analysis.

```python
print("clinical_trials missing values:")
print(clinical_trials.isna().sum())

print("\ndisease_burden missing values:")
print(disease_burden.isna().sum())

print("\nbiotech_funding missing values:")
print(biotech_funding.isna().sum())

print("\ndrug_approvals missing values:")
print(drug_approvals.isna().sum())

print("\npharma_financials missing values:")
print(pharma_financials.isna().sum())
```

## Step 12: Review Data Types

This step reviews the data type of each column. It helps identify whether numerical, categorical, and date columns have been interpreted correctly by Pandas.

```python
print("clinical_trials info:")
clinical_trials.info()

print("\ndisease_burden info:")
disease_burden.info()

print("\nbiotech_funding info:")
biotech_funding.info()

print("\ndrug_approvals info:")
drug_approvals.info()

print("\npharma_financials info:")
pharma_financials.info()
```

## Step 13: Convert Date Columns

The date-related columns are converted from object format to datetime format. This is necessary for time-based analysis, filtering, trend calculations, and chronological visualizations.

```python
clinical_trials["completion_date"] = pd.to_datetime(clinical_trials["completion_date"])
biotech_funding["date"] = pd.to_datetime(biotech_funding["date"])
drug_approvals["approval_date"] = pd.to_datetime(drug_approvals["approval_date"])
```

## Step 14: Validate Date Conversion

This step confirms that the selected date columns were successfully converted to datetime format.

```python
print(clinical_trials["completion_date"].dtype)
print(biotech_funding["date"].dtype)
print(drug_approvals["approval_date"].dtype)
```

## Step 15: Check Duplicate Rows

This step checks whether each dataset contains fully duplicated rows. Duplicate records can distort counts, trends, and aggregated metrics, so they need to be identified before analysis.

```python
print("clinical_trials duplicated rows:", clinical_trials.duplicated().sum())
print("disease_burden duplicated rows:", disease_burden.duplicated().sum())
print("biotech_funding duplicated rows:", biotech_funding.duplicated().sum())
print("drug_approvals duplicated rows:", drug_approvals.duplicated().sum())
print("pharma_financials duplicated rows:", pharma_financials.duplicated().sum())
```

## Initial Data Quality Summary

The initial data quality check shows that all five datasets were successfully loaded and inspected.

Key findings:

- All required CSV files are available in the project folder.
- The datasets contain no missing values.
- No fully duplicated rows were found.
- Date columns were successfully converted to datetime format.
- The datasets are ready for exploratory data analysis.
